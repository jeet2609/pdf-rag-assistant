## 1. Import Required Libraries

```python
from pathlib import Path
import sys

# Add project root to Python path
project_root = Path.cwd().parent
sys.path.append(str(project_root))

from src.embeddings import get_embedding_model
from src.vector_store import load_vector_store
from src.retrieval import create_retriever
from src.rag_chain import (
    load_llm,
    generate_answer_with_sources,
)
```

## 2. Load Embedding Model

```python
embedding_model = get_embedding_model()

print("Embedding model loaded successfully.")
```

## 3. Load FAISS Vector Store

```python
vectore_store = load_vector_store(
	embedding_model=embedding_model
)

print("FAISS vector store loaded successfully.")
```

## 4. Create Retriever

```python
retriever = create_retriever(
	vector_store=vectore_store,
	k=4
)

print("Retriever created successfully.")
```

## 5. Load Gemini

```python
llm = load_llm()

print("Gemini loaded successfully.")
```

## 6. Generate Answer with Sources

```python
question = "What architecture does BERT use?"

result = generate_answer_with_sources(
	question=question,
	retriever=retriever,
	llm=llm
)
```

## 7. Display Answer

```python
print("=" * 80)
print(f"Question: {question}")

print("\nAnswer:")
print(result["answer"])
```

## 8. Display Source Citations

```python
print("\n")
print("=" * 80)
print("Source")
print("=" * 80)

for index, document in enumerate(result["document"], start=1):

	metadata = document.metadata

	print(f"\nSource {index}")

	print(f"Page		: {metadata.get('page', 'N/A')}")
	print(f"Chunk ID	: {metadata.get('chunk_id', 'N/A')}")
	print(f"Source		: {metadata.get('source', 'N/A')}")

	preview = document.page_content[:300].replace("\n", " ")

	print("Context Preview:")

	print(preview)

	print("-" * 80)
```
